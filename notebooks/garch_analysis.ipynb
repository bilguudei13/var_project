# GARCH Stylised Facts and Analysis
This notebook imports the data pipeline functions from `src.var_methods.GARCH` to display the outputs inline without breaking the underlying code structure used for backtesting.

```python
import sys
import os
import pandas as pd

# Add the project root to the path so we can import from src
sys.path.append(os.path.abspath('..'))
from src.var_methods.GARCH import plot_stylised_facts, run_garch_pipeline
```

```python
# Load Risk Factors
factors = pd.read_csv('../data/processed/risk_factors.csv', index_col=0, parse_dates=True)
factors.head()
```

```python
# Plot Stylised Facts and Fit GARCH for each factor
for col in factors.columns:
    print(f"\n{'='*80}\nAnalyzing {col}\n{'='*80}\n")
    series = factors[col].dropna()
    
    # Pass show_plot=True to display it inside the notebook
    plot_stylised_facts(series, col, show_plot=True)
    
    # Fit GARCH model
    res = run_garch_pipeline(series, col)
```
